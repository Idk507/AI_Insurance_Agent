```python
# MSAL Confidential Client Application
def get_msal_app():
    """
    Creates an MSAL confidential client application for authenticating with Azure AD.
    """
    app = msal.ConfidentialClientApplication(
        CLIENT_ID,
        authority=f"https://login.microsoftonline.com/{TENANT_ID}",
        client_credential=CLIENT_SECRET
    )
    return app

# Function to get the Azure access token
def get_access_token():
    """
    Gets an access token from Azure AD using the Confidential Client flow.
    """
    app = get_msal_app()
    
    # Try to acquire token from cache first
    result = app.acquire_token_silent(SCOPES, account=None)
    
    if not result:
        # If the token is not available in the cache, acquire a new one
        result = app.acquire_token_for_client(scopes=SCOPES)
    
    if "access_token" in result:
        return result["access_token"]
    else:
        raise HTTPException(status_code=401, detail="Unable to fetch Azure token")
```
